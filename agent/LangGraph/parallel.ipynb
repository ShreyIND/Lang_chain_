```python
from langgraph.graph import START,END,StateGraph
```

```python
from typing import TypedDict
```

```python
class CricState(TypedDict):
    runs:int
    fours:int
    sixes:int
    balls:int
    sr:float
    runs_in_bound:float
    bpb:float
    summary:str
```

```python
def strikeRate(state:CricState)->CricState:
    runs=state['runs']
    balls=state['balls']
    sr=(runs/balls)*100
    return {'sr':sr}
```

```python
def runs_in_boundaries(state:CricState)->CricState:
    four=state['fours']
    sixes=state['sixes']
    runs=state['runs']
    runs_in_bound=((four*4+sixes*6)/runs)*100
    return state
```

```python
def boundary_per_ball(state:CricState)->CricState:
    balls=state['balls']
    bound=state['fours']+state['sixes']
    state['bpb']=(bound/balls)*100
    return state
```

```python
def summary(state: CricState):

    summary = f"""
    Strike Rate - {state['sr']} \n
    Balls per boundary - {state['bpb']} \n
    Boundary percent - {state['runs_in_bound']}
    """
    
    return {'summary': summary}
```

```python
graph=StateGraph(CricState)
```

```python
graph.add_node(strikeRate)
```

```python
graph.add_node(runs_in_boundaries)
graph.add_node(boundary_per_ball)
graph.add_node(summary)
```

```python
graph.add_edge(START,"runs_in_boundaries")
graph.add_edge(START,"boundary_per_ball")
graph.add_edge(START,"strikeRate")
graph.add_edge("runs_in_boundaries","summary")
graph.add_edge("boundary_per_ball","summary")
graph.add_edge("strikeRate","summary")
graph.add_edge("summary",END)
```

```python
workflow=graph.compile()
```

```python
initial={'runs':100,'sixes':5,'fours':5,'balls':30}
result=workflow.invoke(initial)
```
